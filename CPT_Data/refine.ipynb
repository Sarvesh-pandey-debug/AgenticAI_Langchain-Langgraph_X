```python
import pandas as pd

df = pd.read_excel("old_cptcode.xlsx" , header = 3)
print(df.head())
```

```python
# Sahi sheet specify karein
df1 = pd.read_excel("letest_CPT_code.xlsx", sheet_name="ALL 2026 CPT Codes")
print(df1.head())
```

```python
df1.shape
```

```python
df1 = df1.drop(columns=["Code Status"])
```

```python
df1.shape
```

```python
import pandas as pd
import numpy as np

# 1. Nayi table se 82-83 unique codes nikal kar top ke liye taiyar karein
old_codes = set(df['HCPCS/CPT Code'].dropna().astype(str).str.strip())
df_new_unique = df1[~df1['CPT Codes'].astype(str).str.strip().isin(old_codes)].copy()

df_top = pd.DataFrame()
df_top['Procedure Code Category'] = df_new_unique['Procedure Code Category']
df_top['CPT Codes'] = df_new_unique['CPT Codes']
df_top['Procedure Code Descriptions'] = df_new_unique['Procedure Code Descriptions'] # Naya desc yahan
df_top['Short Description'] = "" # Short desc khali rahega

# 2. Purani table (df) ka data niche ke liye taiyar karein
df_bottom = pd.DataFrame()
df_bottom['Procedure Code Category'] = "" # Category khali rahegi
df_bottom['CPT Codes'] = df['HCPCS/CPT Code']
df_bottom['Procedure Code Descriptions'] = "" # Long desc khali rahega
df_bottom['Short Description'] = df['Short Description'] # Purana desc yahan

# 3. Dono ko combine karein (Naya data sabse upar)
final_df = pd.concat([df_top, df_bottom], ignore_index=True)

# 4. Save to Excel
final_df.to_excel("final_cpt_codes_corrected.xlsx", index=False)

print("Excel file 'final_cpt_codes_corrected.xlsx' save ho gayi hai.")
print("\nExample Format Check (Top Rows):")
print(final_df.head(5))
print("\nExample Format Check (Old Rows):")
print(final_df.iloc[len(df_top):len(df_top)+5])
```
